## Train-Test Split

The dataset has a significant class imbalance, with default cases representing approximately 6.68% of observations.

A stratified train-test split is therefore used to preserve approximately the same target-class distribution in both training and testing datasets.


```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["str","object", "category"]
).columns.tolist()

print("Numerical features:", len(numeric_features))
print("Categorical features:", categorical_features)
```

## Modeling Preprocessing Pipeline

Numerical features are standardized using `StandardScaler`, while categorical features are converted into numerical representations using `OneHotEncoder`.

All preprocessing transformations are fitted only on the training data through a scikit-learn pipeline to prevent data leakage.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])
```

```python
preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])
```

## K-Fold Cross-Validation

To obtain a more reliable estimate of model performance, Stratified 5-Fold Cross-Validation will be performed on the training dataset.

Stratification is used because the target variable is highly imbalanced. Each fold will therefore maintain approximately the same proportion of default and non-default observations.

The test set will remain completely untouched during cross-validation and will only be used for the final evaluation of the selected model.

```python
from sklearn.model_selection import StratifiedKFold, cross_validate
```

```python
skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc",
    "pr_auc": "average_precision"
}
```

```python
def cross_validate_model(model, X, y, model_name):
    
    cv_results = cross_validate(
        model,
        X,
        y,
        cv=skf,
        scoring=scoring,
        n_jobs=-1
    )
    
    return {
        "Model": model_name,
        "Accuracy": cv_results["test_accuracy"].mean(),
        "Precision": cv_results["test_precision"].mean(),
        "Recall": cv_results["test_recall"].mean(),
        "F1": cv_results["test_f1"].mean(),
        "ROC-AUC": cv_results["test_roc_auc"].mean(),
        "PR-AUC": cv_results["test_pr_auc"].mean(),
        
        "Accuracy_STD": cv_results["test_accuracy"].std(),
        "Precision_STD": cv_results["test_precision"].std(),
        "Recall_STD": cv_results["test_recall"].std(),
        "F1_STD": cv_results["test_f1"].std(),
        "ROC-AUC_STD": cv_results["test_roc_auc"].std(),
        "PR-AUC_STD": cv_results["test_pr_auc"].std()
    }
```

## Logistic Regression with Class Weighting

The target variable is highly imbalanced, with approximately 93.32% non-default and 6.68% default observations.

To reduce the effect of this imbalance, Logistic Regression is trained using `class_weight="balanced"`.

This automatically assigns greater weight to the minority default class during model training without changing the original training data distribution.

```python
balanced_logistic_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ))
])
```

The baseline Logistic Regression achieved 93.58% accuracy but had a low recall of 16.06% for the default class. This indicates that the model is highly conservative in identifying potential defaulters.

## XGBoost Classifier

XGBoost is introduced as a boosting-based classification model to capture complex nonlinear relationships and interactions between features.

Unlike Random Forest, which builds trees independently and combines their predictions, XGBoost builds trees sequentially, with each new tree focusing on correcting errors made by previous trees.

Because the target variable is highly imbalanced, class weighting is incorporated through the `scale_pos_weight` parameter.

The model will be evaluated primarily using recall, false negatives, F1-score, ROC-AUC, and PR-AUC.

```python
scale_pos_weight = (
    (y_train == 0).sum() /
    (y_train == 1).sum()
)

scale_pos_weight
```

```python
from xgboost import XGBClassifier

xgb_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", XGBClassifier(
        n_estimators=300,
        max_depth=5,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=scale_pos_weight,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    ))
])
```

## Classification Threshold Optimization

The default classification threshold of 0.5 is not necessarily optimal for credit-risk assessment.

Since the primary business objective is to minimize false negatives and identify as many potential defaulters as possible, different probability thresholds are evaluated for all candidate models.

Lowering the threshold generally increases recall and reduces false negatives, but also increases false positives.

The threshold will therefore be evaluated based on recall, false negatives, precision, F1-score, and the resulting business trade-off.

```python
thresholds = [0.50, 0.45, 0.40, 0.35, 0.30, 0.25, 0.20, 0.15, 0.10]
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

def evaluate_thresholds(y_true, y_prob, model_name, thresholds):
    
    results = []

    for threshold in thresholds:
        
        y_pred_threshold = (y_prob >= threshold).astype(int)
        
        tn, fp, fn, tp = confusion_matrix(
            y_true,
            y_pred_threshold
        ).ravel()
        
        results.append({
            "Model": model_name,
            "Threshold": threshold,
            "Accuracy": accuracy_score(y_true, y_pred_threshold),
            "Precision": precision_score(
                y_true, y_pred_threshold, zero_division=0
            ),
            "Recall": recall_score(
                y_true, y_pred_threshold, zero_division=0
            ),
            "F1": f1_score(
                y_true, y_pred_threshold, zero_division=0
            ),
            "False_Positives": fp,
            "False_Negatives": fn,
            "True_Positives": tp,
            "True_Negatives": tn
        })

    return pd.DataFrame(results)
```

```python
threshold_results = pd.concat([
    evaluate_thresholds(
        y_test,
        y_prob,
        "Baseline Logistic",
        thresholds
    ),
    
    evaluate_thresholds(
        y_test,
        y_prob_balanced,
        "Balanced Logistic",
        thresholds
    ),
    
    evaluate_thresholds(
        y_test,
        y_prob_rf,
        "Random Forest",
        thresholds
    ),
    
    evaluate_thresholds(
        y_test,
        y_prob_xgb,
        "XGBoost",
        thresholds
    )
], ignore_index=True)
```

```python
threshold_results.round(4)
```

```python
high_recall = threshold_results[
    threshold_results["Recall"] >= 0.80
]

high_recall[
    [
        "Model",
        "Threshold",
        "Recall",
        "Precision",
        "F1",
        "False_Negatives",
        "False_Positives"
    ]
].sort_values(
    ["Model", "Threshold"],
    ascending=[True, False]
).round(4)
```
